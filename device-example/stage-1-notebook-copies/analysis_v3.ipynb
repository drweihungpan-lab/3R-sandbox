# Device usage: when does each device reach its limit?
Daily measurement counts for three devices. Each device can take 100,000 measurements in total.

Last 180 days, and leave out days with zero measurements (maintenance days).

```python
import numpy as np
import pandas as pd

LIMIT = 100_000
df = pd.read_csv("data/device_counts.csv", parse_dates=["date"])
```

```python
for device, g in df.groupby("device_id"):
    g = g.sort_values("date")
    g = g[g["daily_count"] > 0]
    g = g.assign(total=g["daily_count"].cumsum()).tail(180)
    x = np.arange(len(g))
    y = g["total"].to_numpy()
    slope, intercept = np.polyfit(x, y, 1)
    days = int(np.ceil((LIMIT - intercept) / slope))
    print(device, "reaches the limit on", (g["date"].iloc[0] + pd.Timedelta(days=days)).date())
```
