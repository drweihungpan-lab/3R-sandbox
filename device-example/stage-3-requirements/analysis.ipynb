# Device usage: when does each device reach its limit?

Dependencies are declared in `requirements.txt`. Run the cells top to bottom.

```python
from loader import load_all
from model import limit_dates
```

```python
df = load_all()
```

```python
# Turn each day's count into a running total.
df["daily_count"] = df.groupby("device_id")["daily_count"].cumsum()
```

```python
# 10,000 measurements arrived late and belong to 1 December. Add them to the totals.
df.loc[df["date"] >= "2025-12-01", "daily_count"] += 10_000
```

```python
for device, date in limit_dates(df).items():
    print(device, "reaches the limit on", date)
```
